```python
def analyze_sentiment(text):
  sentiment_score = sia.polarity_scores(text)
  if sentiment_score['compound'] >= 0.05:
    return "Positve Sentence"
  elif sentiment_score['compound'] <= -0.05:
    return "Negative Sentence"
  else:
    return "Neutral Sentence"
```

```python
def chatbot_app():
  print("Hello! I'm your chatbot. (Type 'bye' to exit.)")
  while True:
    user_input = input("You: ")

    # -- Sentiment Analysis Part --
    if user_input.lower() == "bye":
      print("ChatBot: Goodbye! Have a great day!")
      break
    elif "sentiment" in user_input:
      print("Chatbot: Tell me a sentence to analyze.")
      sentiment_input = input("You: ")
      print("Chatbot: Sentiment Analysis Result -", analyze_sentiment(sentiment_input))
    else:
    # -- End of Sentiment Analysis Part --

      response = chatbot.respond(user_input.lower())
      if response:
        print("Chatbot:", response)
        if "bye" in user_input.lower():
          break
      else:
        print("Chatbot: I'm not sure how to respond to that.")
```

```python
chatbot_app()
```
